# [dbo].[Admin_GetAllPages]

```sql
/****** Object:  StoredProcedure [dbo].[Admin_GetAllPages]    Script Date: 02-Jan-26 13:29:27 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
-- =============================================
-- ADMIN PAGE PROCEDURES
-- =============================================

-- 1. Get all pages for admin list
CREATE   PROCEDURE [dbo].[Admin_GetAllPages]
AS
BEGIN
    SET NOCOUNT ON;
    SELECT Id, Slug, IsPublished, SortOrder
    FROM Pages 
    WHERE IsMenuItem = 1 OR Slug = 'home'  -- Show menu + home
    ORDER BY SortOrder, Slug;
END
GO
```

# [dbo].[Admin_GetPageForEdit]

```sql
/****** Object:  StoredProcedure [dbo].[Admin_GetPageForEdit]    Script Date: 02-Jan-26 13:29:27 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

-- 2. Get page for editing (page + all language contents)
CREATE   PROCEDURE [dbo].[Admin_GetPageForEdit]
    @PageId INT
AS
BEGIN
    SET NOCOUNT ON;
    
    -- Page metadata
    SELECT Id, Slug, IsPublished, IsMenuItem, MenuSortOrder, SortOrder
    FROM Pages 
    WHERE Id = @PageId;

    -- Language contents
    SELECT 
        pc.Id, pc.LanguageId, l.Code, l.Name,
        pc.SeoTitle, pc.SeoDescription,
        pc.HeroTitle, pc.HeroSubtitle, pc.HeroCtaText, pc.HeroCtaUrl,
        ISNULL(pc.ContentSections, '[]') as ContentSections,
        pc.IsPublished
    FROM PageContents pc
    INNER JOIN Languages l ON pc.LanguageId = l.Id
    WHERE pc.PageId = @PageId;
END
GO
```

# [dbo].[Admin_SavePage]

```sql
/****** Object:  StoredProcedure [dbo].[Admin_SavePage]    Script Date: 02-Jan-26 13:29:27 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

-- 3. Save page (upsert page + language contents)
CREATE   PROCEDURE [dbo].[Admin_SavePage]
    @PageId INT,
    @Slug NVARCHAR(100),
    @IsPublished BIT,
    @IsMenuItem BIT,
    @MenuSortOrder INT = NULL,
    @SortOrder INT = 0,
    @LanguageContents NVARCHAR(MAX)  -- JSON array of language data
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @LanguageData TABLE (
        LanguageId TINYINT,
        SeoTitle NVARCHAR(200) NULL,
        SeoDescription NVARCHAR(500) NULL,
        HeroTitle NVARCHAR(200) NULL,
        HeroSubtitle NVARCHAR(500) NULL,
        HeroCtaText NVARCHAR(100) NULL,
        HeroCtaUrl NVARCHAR(500) NULL,
        ContentSections NVARCHAR(MAX),
        IsPublished BIT
    );

    -- Parse JSON language contents
    INSERT INTO @LanguageData
    SELECT 
        JSON_VALUE(value, '$.LanguageId'),
        JSON_VALUE(value, '$.SeoTitle'),
        JSON_VALUE(value, '$.SeoDescription'),
        JSON_VALUE(value, '$.HeroTitle'),
        JSON_VALUE(value, '$.HeroSubtitle'),
        JSON_VALUE(value, '$.HeroCtaText'),
        JSON_VALUE(value, '$.HeroCtaUrl'),
        ISNULL(JSON_VALUE(value, '$.ContentSections'), '[]'),
        JSON_VALUE(value, '$.IsPublished')
    FROM OPENJSON(@LanguageContents);

    BEGIN TRANSACTION;

    BEGIN TRY
        -- Update page metadata
        UPDATE Pages
        SET Slug = @Slug,
            IsPublished = @IsPublished,
            IsMenuItem = @IsMenuItem,
            MenuSortOrder = @MenuSortOrder,
            SortOrder = @SortOrder,
            UpdatedAt = GETUTCDATE()
        WHERE Id = @PageId;

        -- Upsert language contents
        MERGE PageContents AS target
        USING @LanguageData AS source
        ON (target.PageId = @PageId AND target.LanguageId = source.LanguageId)
        WHEN MATCHED THEN
            UPDATE SET
                SeoTitle = source.SeoTitle,
                SeoDescription = source.SeoDescription,
                HeroTitle = source.HeroTitle,
                HeroSubtitle = source.HeroSubtitle,
                HeroCtaText = source.HeroCtaText,
                HeroCtaUrl = source.HeroCtaUrl,
                ContentSections = source.ContentSections,
                IsPublished = source.IsPublished,
                PublishedAt = CASE WHEN source.IsPublished = 1 THEN GETUTCDATE() ELSE target.PublishedAt END,
                UpdatedAt = GETUTCDATE()
        WHEN NOT MATCHED BY TARGET THEN
            INSERT (PageId, LanguageId, SeoTitle, SeoDescription, HeroTitle, HeroSubtitle, 
                    HeroCtaText, HeroCtaUrl, ContentSections, IsPublished, PublishedAt, CreatedAt)
            VALUES (@PageId, source.LanguageId, source.SeoTitle, source.SeoDescription, 
                    source.HeroTitle, source.HeroSubtitle, source.HeroCtaText, source.HeroCtaUrl, 
                    source.ContentSections, source.IsPublished, 
                    CASE WHEN source.IsPublished = 1 THEN GETUTCDATE() ELSE NULL END, GETUTCDATE());

        COMMIT TRANSACTION;
        SELECT 1 as Success;
    END TRY
    BEGIN CATCH
        ROLLBACK TRANSACTION;
        THROW;
    END CATCH
END
GO
```

# [dbo].[GetPageBySlug]

```sql
/****** Object:  StoredProcedure [dbo].[GetPageBySlug]    Script Date: 02-Jan-26 13:29:27 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
CREATE   PROCEDURE [dbo].[GetPageBySlug]
    @Slug NVARCHAR(100),
    @LanguageCode NVARCHAR(5)
AS
BEGIN
    SET NOCOUNT ON;
    
    DECLARE @LanguageId TINYINT = (SELECT Id FROM Languages WHERE Code = @LanguageCode AND IsActive = 1);
    IF @LanguageId IS NULL
        SET @LanguageId = (SELECT Id FROM Languages WHERE IsDefault = 1);

    SELECT 
        p.Id as PageId,
        p.Slug,
        l.Code as LanguageCode,
        pc.SeoTitle,
        pc.SeoDescription,
        pc.HeroTitle,
        pc.HeroSubtitle,
        pc.HeroCtaText,
        pc.HeroCtaUrl,
        pc.ContentSections,
        pc.OpenGraphTitle,
        pc.OpenGraphDescription,
        pc.OpenGraphImage
    FROM Pages p
    INNER JOIN PageContents pc ON p.Id = pc.PageId AND pc.LanguageId = @LanguageId
    INNER JOIN Languages l ON pc.LanguageId = l.Id
    WHERE p.Slug = @Slug AND p.IsPublished = 1 AND pc.IsPublished = 1;

    SELECT TOP 6 
        pr.Id,
        pr.Slug,
        pr.ImageUrl,
        pr.DurationHours,
        pr.MaxParticipants,
        prt.Title,
        prt.ShortDescription,
        ctT.Name as CategoryName
    FROM PagePrograms pp
    INNER JOIN Programs pr ON pp.ProgramId = pr.Id AND pr.IsPublished = 1
    INNER JOIN ProgramTranslations prt ON pr.Id = prt.ProgramId AND prt.LanguageId = @LanguageId
    LEFT JOIN ProgramCategories ct ON pr.CategoryId = ct.Id
    LEFT JOIN ProgramCategoryTranslations ctT ON ct.Id = ctT.CategoryId AND ctT.LanguageId = @LanguageId
    WHERE pp.PageId = (SELECT Id FROM Pages WHERE Slug = @Slug)
    ORDER BY pp.SortOrder;
END
GO
```
